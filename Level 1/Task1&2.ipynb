```python
# TASK 1: DATA CLEANING AND PREPROCESSING
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
%matplotlib inline
```

```python
sns.set_theme(style="whitegrid")
```

```python
# Dataset from the shared google Drive
df = pd.read_csv('churn-bigml-80.csv')
```

```python
print("--- Data Info ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Summary Statistics ---")
display(df.describe())
```

```python
df_cleaned = df.dropna()
df_cleaned = df_cleaned.drop_duplicates()
print("\n--- Cleaning Complete ---")
print(f"Cleaned data shape (rows, columns): {df_cleaned.shape}")
print("-" * 40)
```

```python
#TASK 2 EXPLORATORY DATA ANALYSIS (EDA)
```

```python
# 1. Calculate remaining summary statistics & corelations b/w numerical
print("\n--- Mode of the dataset ---")
display(df_cleaned.mode().iloc[0])
numeric_cols = df_cleaned.select_dtypes(include=['float64', 'int64'])
print("\n--- Correlation Matrix ---")
display(numeric_cols.corr())
```

```python
#Histogram : Distribution of Customer Service Calls
plt.figure(figsize=(8, 4))
sns.histplot(df_cleaned['Customer service calls'], bins=10, kde=False, color='skyblue')
plt.title('Distribution of Customer Service Calls')
plt.xlabel('Number of Calls')
plt.ylabel('Frequency')
plt.show()

#Scatter Plot : Seeing the correlation between Minutes and Charge
plt.figure(figsize=(8, 4))
sns.scatterplot(x='Total day minutes', y='Total day charge', data=df_cleaned, hue='Churn')
plt.title('Scatter Plot: Total Day Minutes vs Total Day Charge')
plt.show()

# Boxplot: Comparing Total Day Minutes between customers who churned and those who stayed
plt.figure(figsize=(8, 4))
sns.boxplot(x='Churn', y='Total day minutes', data=df_cleaned, palette='Set2')
plt.title('Boxplot: Total Day Minutes by Churn')
plt.show()
```
